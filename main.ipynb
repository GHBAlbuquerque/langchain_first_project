```python
from langchain_core.prompts import ChatPromptTemplate #formats the user input into a structured prompt for the model
from langchain.chat_models import init_chat_model #used to call any AI model without specific configs
from langchain_core.output_parsers import StrOutputParser #extract model answer and makes it redable for langchain
from langchain_core.runnables import RunnableLambda #create a function

from dotenv import load_dotenv

load_dotenv()

prompt = ChatPromptTemplate.from_template("Explain using 2 sentences about {subject}")
model = init_chat_model(model="gpt-5-nano", model_provider="openai") #autodetects api_key with load_dotenv
parser = StrOutputParser()

pipeline = prompt | model | parser #operator overloading done by langchain -> run A, feed into B, then C
response = pipeline.invoke({"subject": "quantum physics"})
print(response)
```

```python
def capitalize(text: str) -> str:
    return text.upper()

extra_step = RunnableLambda(capitalize)

pipeline = prompt | model | parser | extra_step

response = pipeline.invoke({"subject": "video games"})
print(response)
```
