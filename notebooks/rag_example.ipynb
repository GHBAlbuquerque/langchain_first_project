```python
from langchain_core.documents import Document
from langchain_openai import OpenAIEmbeddings
from langchain_core.vectorstores import InMemoryVectorStore

docs = [
    Document(
        page_content="""Welcome to Nexus Corp. Founded in 2035, Nexus Corp is the leading manufacturer of commercial anti-gravity propulsion systems. 
Our mission is to make deep space travel accessible to the commercial sector. This handbook outlines the core operational guidelines, 
benefits, and safety protocols for all engineering and support staff working at the Lunar Base Alpha facility.""",
        metadata={
            "source": "employee_handbook_v2.pdf",
            "page": 1,
            "category": "Company Overview",
            "author": "HR Department"
        }
    ),
    Document(
        page_content="""The core technology powering our vessels is the Quantum Flux Drive (QFD). The QFD operates by generating a localized 
gravitational anomaly, allowing the ship to bypass standard mass restrictions. Engineers must ensure that the QFD's beryllium 
containment sphere is kept at a stable temperature of exactly 4 Kelvin. Any temperature fluctuations above 10 Kelvin will trigger 
an automatic system shutdown to prevent structural micro-fractures.""",
        metadata={
            "source": "employee_handbook_v2.pdf",
            "page": 2,
            "category": "Technical Specifications",
            "author": "Engineering Division"
        }
    ),
    Document(
        page_content="""Standard operating procedure for QFD maintenance requires weekly diagnostics. Maintenance personnel must wear 
Class-A radiation suits when entering the engine bay. The diagnostic sequence involves checking the coolant flow rates, 
calibrating the magnetic localized field, and purging the exhaust manifold of any residual plasma buildup. Maintenance logs 
must be submitted to the central database by 17:00 standard lunar time every Friday.""",
        metadata={
            "source": "employee_handbook_v2.pdf",
            "page": 3,
            "category": "Maintenance Protocols",
            "author": "Engineering Division"
        }
    ),
    Document(
        page_content="""Safety is our top priority at Nexus Corp. In the event of a containment breach in the QFD, the engine bay will 
automatically seal and initiate a hard vacuum purge. Personnel have exactly 30 seconds to evacuate the bay upon hearing 
the primary alarm. Do not attempt to manually override the blast doors during a lockdown sequence. Familiarize yourself 
with the locations of emergency oxygen reserves, which are located every 50 meters in the main corridors.""",
        metadata={
            "source": "employee_handbook_v2.pdf",
            "page": 4,
            "category": "Safety Protocols",
            "author": "Security & Safety Council"
        }
    ),
    Document(
        page_content="""Nexus Corp offers a comprehensive benefits package. Employees stationed at Lunar Base Alpha receive a 25% hazard pay 
bonus, comprehensive medical coverage including micro-gravity bone density treatments, and 30 days of paid terrestrial leave 
per solar year. To request terrestrial leave, employees must submit Form T-42 to their department supervisor at least 60 days 
in advance. Travel to and from Earth is complimentary on company shuttles.""",
        metadata={
            "source": "employee_handbook_v2.pdf",
            "page": 5,
            "category": "Employee Benefits",
            "author": "HR Department"
        }
    )
]

embedding = OpenAIEmbeddings(model="text-embedding-3-large")
vector_store = InMemoryVectorStore.from_documents(documents=docs, embedding=embedding)
retriever = vector_store.as_retriever(search_kwargs={"k":2})

queries = [
    "What is the required operating temperature for the Quantum Flux Drive's containment sphere, and what happens if it exceeds 10 Kelvin?",
    "I am assigned to QFD maintenance this week. What protective gear do I need to wear, and what are the specific steps in the diagnostic sequence?",
    "What are the specific medical and financial benefits of being stationed at Lunar Base Alpha, and what form must be submitted to request time off on Earth?",
    "If a containment breach happens in the engine bay, how much time do I have to evacuate, and where exactly can I find emergency oxygen in the corridors?",
    "How does the company's founding mission relate to the specific technology powering the ships, and what time must the weekly maintenance logs for that technology be submitted?"
]

for query in queries:
    print("=" * 60)
    print("Query:", query)
    results = retriever.invoke(query)
    for i, doc in enumerate(results, start=1):
        print(f"Result {i}: {doc.page_content}")


# Example of how you might verify the documents in your app:
print(f"Loaded {len(docs)} documents.")
print(f"Sample Metadata from Doc 1: {docs[0].metadata}")
```
